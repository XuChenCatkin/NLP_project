```python
import json
import csv

def format_percent(value):
    """Convert float to percentage string with 4 significant figures."""
    return f"{value * 100:.4g}"

def json_to_csv(json_file, csv_file):
    with open(json_file, 'r') as f:
        data = json.load(f)

    rows = []
    # Support both single item and list of items
    if isinstance(data, dict):
        data = [data]

    for item in data:
        dataset = item.get("data_set", "")
        method = item.get("generation_model", "")
        perf = item.get("performance", {})

        sas = format_percent(perf.get("sas", {}).get("mean_similarity", 0.0))
        gpt_score = format_percent(perf.get("GPT Score", {}).get("average GPTscore", 0.0))
        precision = format_percent(perf.get("bertscore", {}).get("precision", 0.0))
        recall = format_percent(perf.get("bertscore", {}).get("recall", 0.0))
        f1 = format_percent(perf.get("bertscore", {}).get("f1", 0.0))

        rows.append([dataset, method, sas, gpt_score, precision, recall, f1])

    # Write to CSV
    with open(csv_file, 'w', newline='') as f:
        writer = csv.writer(f)
        writer.writerow(["dataset", "method", "sas", "GPT Score", "precision", "recall", "f1"])
        writer.writerows(rows)



```
