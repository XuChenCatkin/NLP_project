```python
import json
import random
# path
#data/QA/multi_question.json
# Load the data from the file
with open('multi_question.json', 'r') as f:
    data = json.load(f)

# Shuffle the data
#random.shuffle(data)

# Re-label the 'id' for each item
for index, item in enumerate(data):
    item['id'] = index + 1
    item['category'] = 'multi_labeled'
# Get the total number of questions
total_questions = len(data)

# Print the total number of questions
print(f"Total number of questions: {total_questions}")

# save the modified data back to the file
with open('multi_question.json', 'w') as f:
    json.dump(data, f, indent=2)
print("File has been shuffled, re-labeled, and saved successfully!")    
# # Print the shuffled and re-labeled data
# print(json.dumps(data, indent=2))
```
